```python
remoteok["salary_min"] = remoteok["salary"].str.extract(r"(\d[\d,]*)\s*[-–]").astype(int)
remoteok["salary_max"] = remoteok["salary"].str.extract(r"[-–]\s*(\d[\d,]*)").astype(int)

remoteok["salary_avg"] = remoteok[["salary_min", "salary_max"]].mean(axis=1)
remoteok = remoteok.drop(columns=["salary_min", "salary_max", "salary"])  # drop unwanted columns
mean_salary = int(remoteok.loc[remoteok["salary_avg"] > 0, "salary_avg"].mean())
remoteok["salary"] = remoteok["salary_avg"].replace(0, mean_salary).astype(int)
remoteok = remoteok.drop(columns = ["salary_avg"])
```

```python
remotely.dropna(subset=["title"], inplace=True)
```

```python
import re
remotely["salary"] = remotely["categories"].str.extract((r'(\$\d{1,3}(?:,\d{3})*(?:\s*-\s*\$\d{1,3}(?:,\d{3})*)?\s*(?:or more)?\s*USD)'))
remotely["salary"] = remotely["salary"].fillna("Not mention")
```

```python
def extract_location_list_remotely(text):
    # handle NaN or non-string values
    if not isinstance(text, str) or text.strip() == "":
        return []

    # remove emojis (flags etc.)
    # text = re.sub(r'[^\w\s,]', '', text)

    # try to get everything after 'USD' (if salary exists)
    match = re.search(r'USD\s*(.*)', text)
    if match:
        loc_text = match.group(1)
    else:
        # otherwise, after 'Full-Time' or 'Contract'
        match2 = re.search(r'(?:Full-Time|Contract)\s*(.*)', text)
        loc_text = match2.group(1) if match2 else ''
    
    loc_text = loc_text.strip()

    loc_text = re.sub(r'[^\w\s,]', '',loc_text)

    if not loc_text:
        return []

    # handle special case
    if 'Anywhere in the World' in loc_text:
        return ['Anywhere in the World']

    # split by spaces and clean up
    locs = [x.strip() for x in loc_text.split() if x.strip()]
    return locs
```

```python
remotely['location'] = remotely['categories'].apply(extract_location_list_remotely)
```
