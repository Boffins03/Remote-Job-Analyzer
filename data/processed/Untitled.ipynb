```python
import re
remotely["salary"] = remotely["categories"].str.extract((r'(\$\d{1,3}(?:,\d{3})*(?:\s*-\s*\$\d{1,3}(?:,\d{3})*)?\s*(?:or more)?\s*USD)'))
# remotely["salary"] = remotely["salary"].fillna("Not mention")
```

```python
def extract_location_list_remotely(text):
    # handle NaN or non-string values
    if not isinstance(text, str) or text.strip() == "":
        return []

    # remove emojis (flags etc.)
    # text = re.sub(r'[^\w\s,]', '', text)

    # try to get everything after 'USD' (if salary exists)
    match = re.search(r'USD\s*(.*)', text)
    if match:
        loc_text = match.group(1)
    else:
        # otherwise, after 'Full-Time' or 'Contract'
        match2 = re.search(r'(?:Full-Time|Contract)\s*(.*)', text)
        loc_text = match2.group(1) if match2 else ''
    
    loc_text = loc_text.strip()

    loc_text = re.sub(r'[^\w\s,]', '',loc_text)

    if not loc_text:
        return []

    # handle special case
    if 'Anywhere in the World' in loc_text:
        return ['Anywhere in the World']

    # split by spaces and clean up
    locs = [x.strip() for x in loc_text.split() if x.strip()]
    return locs
```

```python
remotely['location'] = remotely['categories'].apply(extract_location_list_remotely)
```

```python
# ---- Function to clean and average salary ----
def clean_salary(value):
    if not isinstance(value, str):
        return None
    value = value.replace(",", "")
    
    # Match salary range: e.g. $25000 - $48999
    range_match = re.findall(r"\$(\d+)", value)
    
    if len(range_match) == 2:
        # Take average of two numbers
        low, high = map(int, range_match)
        return int((low + high) / 2)
    
    elif len(range_match) == 1:
        # Handle "$100000 or more USD"
        return int(range_match[0])
    
    else:
        # Handle "Not mention" or missing
        return None

# Apply the function
remotely["salary_cleaned"] = remotely["salary"].apply(clean_salary).astype("Int64")

# Optional: Fill missing with mean
mean_salary = int(remotely["salary_cleaned"].mean())
# remotely["salary_cleaned"].fillna(mean_salary, inplace=True)
remotely.fillna({'salary_cleaned':mean_salary},inplace=True)
```

```python
# Function to clean location column
def clean_location(loc_list):
    if 'Anywhere in the World' in loc_list:
        return 'Remote'
    else:
        return ', '.join(loc_list)  # join multiple countries as string if needed

# Apply the function
remotely['location_cleaned'] = remotely['location'].apply(clean_location)
```

```python
remotely = remotely.drop(columns=["company_location", "categories","salary","location"])
remotely["salary"] = remotely["salary_cleaned"]
remotely["location"] = remotely["location_cleaned"]
remotely = remotely.drop(columns = ["salary_cleaned","location_cleaned"])
```

```python
# Extract salary range like $120k – $180k
wellfound["salary"] = wellfound["details"].str.extract(r"(\$[\d,]+k\s*[–-]\s*\$[\d,]+k)")
```

```python
# Extract text after '•' (dot symbol) or after 'Remote'
wellfound["location"] = wellfound["details"].str.extract(r"(?:•\s*|Remote\s*•\s*)([A-Za-z\s\+\d]+)")
```

```python
# 1. FIX UNICODE AND REMOVE NOISE
wellfound["clean"] = (
    wellfound["details"]
    .str.replace("â€“", "-", regex=False)
    .str.replace("â€”", "-", regex=False)
    .str.replace("â€¢", "•", regex=False)
    .str.replace("No equity", "", regex=False)
)

# remove +XX years of exp or +1, +2, etc
wellfound["clean"] = wellfound["clean"].str.replace(r"\+\d+.*", "", regex=True)

# remove equity percentage ranges like "0.0% – 0.006%"
wellfound["clean"] = wellfound["clean"].str.replace(r"\d+\.\d+%\s*-\s*\d+\.\d+%", "", regex=True)

# remove "• " duplicates or leftover dots
wellfound["clean"] = wellfound["clean"].str.replace(r"•\s*•", "•", regex=True)


# 2. EXTRACT SALARY
# wellfound["salary"] = wellfound["clean"].str.extract(r"(\$\d+[kK]?\s*-\s*\$\d+[kK]?)")


# 3. EXTRACT WORK TYPE
wellfound["work_type"] = wellfound["clean"].str.extract(
    r"(Remote only|Remote|Onsite only|Onsite or remote|Onsite|Hybrid)",
    expand=False
)


# 4. EXTRACT LOCATION (ROBUST & FIXED)
wellfound["location_raw"] = wellfound["clean"].str.extract(
    r"(?:Remote only|Remote|Onsite only|Onsite or remote|Onsite|Hybrid)"
    r"\s*[-•]?\s*([A-Za-z\s]+)",
    expand=False
)

# clean city/country
wellfound["location_raw"] = wellfound["location_raw"].str.strip()


# 5. BUILD FINAL LOCATION COLUMN
def build_location(row):
    wt = row["work_type"]
    loc = row["location_raw"]

    if pd.notna(wt) and pd.notna(loc):
        return f"{wt} , {loc}"
    if pd.notna(wt):
        return wt
    if pd.notna(loc):
        return loc
    return None

wellfound["location_final"] = wellfound.apply(build_location, axis=1)
```
